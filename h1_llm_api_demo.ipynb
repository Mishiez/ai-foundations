# H1 — LLM API Demonstration
This notebook demonstrates a working Python environment, secure API-key
handling, an LLM API request, and a structured response.

```python
import os
from dotenv import load_dotenv
from google import genai
```

```python
load_dotenv()
api_key = os.getenv("GEMINI_API_KEY")
print("API key loaded:", bool(api_key))
```

```python
client = genai.Client(api_key=api_key)
```

```python
response = client.models.generate_content(
    model="gemini-3.6-flash",
    contents="Explain machine learning to a beginner. Return a title, a summary, and three key points."
)
```

```python
print(response.text)
```
